Aditi Verma (C01) and Sneha Joshi (C05) are both from Delhi. Using the orders and products data, build two sets
— one containing the product Categories Aditi has ordered, and one containing the Categories Sneha has ordered.
Use the set intersection operator (&) to find categories they have both shopped in. Print the result.

```python
products_list = {p["Product_ID"]: p["Category"] for p in products}

aditi_categories = set()
sneha_categories = set()

for x in orders:
  if x["Customer_ID"] == "C01":
    aditi_categories.add(products_list[x["Product_ID"]])
  elif x["Customer_ID"] == "C05":
    sneha_categories.add(products_list[x["Product_ID"]])

common_categories = aditi_categories & sneha_categories
print(f"Common Categories shopped by Aditi & Sneha: {common_categories}")
```


Calculate Total Revenue by adding up the Order_Value of every order — but while looping, use continue to skip
any order whose Status is "Cancelled", since cancelled orders must never count as revenue. Print the final Total
Revenue.



```python
total_revenue = 0

for order in orders:
    if order['Status'] == 'Cancelled':
        continue
    for product in products:
        if order['Product_ID'] == product['Product_ID']:
            order_value = product['Price'] * order['Quantity']
            total_revenue += order_value

print(f"Total Revenue: {total_revenue}")
```

For every order that is NOT cancelled, classify it using if–elif–else as: "High Value" (Order_Value > 3000),
"Medium Value" (1000 to 3000), or "Low Value" (below 1000). Print each Order_ID with its classification.

```python
product_prices = {product['Product_ID']: product['Price'] for product in products}

for order in orders:
    if order['Status'] == 'Cancelled':
        continue
    product_id = order['Product_ID']
    quantity = order['Quantity']
    order_value = product_prices.get(product_id, 0) * quantity

    if order_value > 3000:
        classification = 'High Value'
    elif 1000 <= order_value <= 3000:
        classification = 'Medium Value'
    else:
        classification = 'Low Value'

    print(f"Order ID: {order['Order_ID']}, Classification: {classification}")
```

Loop through all orders and count how many used each Payment_Mode ("UPI", "Credit Card", "Debit Card").
Print the three counts.

```python
upi_count = 0
credit_card_count = 0
debit_card_count = 0

for order in orders:
    if order['Payment_Mode'] == 'UPI':
        upi_count += 1
    elif order['Payment_Mode'] == 'Credit Card':
        credit_card_count += 1
    elif order['Payment_Mode'] == 'Debit Card':
        debit_card_count += 1

print(f"UPI Count: {upi_count}")
print(f"Credit Card Count: {credit_card_count}")
print(f"Debit Card Count: {debit_card_count}")
```

Write a function get_product_price(product_id, products_list) that returns the Price of the matching product, or
None if the ID is not found.

```python
def get_product_price(product_id, products_list):
    for product in products_list:
        if product['Product_ID'] == product_id:
            return product['Price']
    return None
```
